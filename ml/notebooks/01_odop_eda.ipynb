```python
import pandas as pd

df = pd.read_csv(
    "../datasets/raw/odop.csv",
    encoding="cp1252"
)

print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("States:", df["State"].nunique())
print("Districts:", df["District"].nunique())
print("Products:", df["Product"].nunique())
print("Categories:", df["Category"].nunique())
print("Sectors:", df["Sector"].nunique())
```

```python
df["Sector"].value_counts()
```

```python
handicraft_df = df[
    df["Sector"].str.contains("Handicraft", case=False, na=False)
]

print("Handicraft rows:", len(handicraft_df))
handicraft_df.head(10)
```

```python
# Normalize text columns
handicraft_df = df.copy()

for col in ["State", "Product", "District", "Category", "Sector",
            "Description", "GI Status", "Ministry/ Department"]:
    handicraft_df[col] = handicraft_df[col].astype(str).str.strip()

# Normalize Sector
handicraft_df["Sector"] = handicraft_df["Sector"].str.lower()

# Keep only handicraft records
handicraft_df = handicraft_df[
    handicraft_df["Sector"] == "handicraft"
].copy()

print("Handicraft rows after cleaning:", len(handicraft_df))
```

```python
handicraft_df["State"].value_counts()
handicraft_df["GI Status"].value_counts()
handicraft_df["Product"].value_counts().head(20)
```

```python
print("===== STATES =====")
print(handicraft_df["State"].value_counts())

print("\n===== GI STATUS =====")
print(handicraft_df["GI Status"].value_counts())

print("\n===== TOP 20 PRODUCTS =====")
print(handicraft_df["Product"].value_counts().head(20))
```

```python
print("Unique handicraft products:", handicraft_df["Product"].nunique())

print("\nAll product names:")
print(sorted(handicraft_df["Product"].unique()))
```

```python
print("===== GI STATUS =====")

gi_counts = handicraft_df["GI Status"].value_counts()

print(gi_counts)
print("\nPercentage:")
print((gi_counts / len(handicraft_df) * 100).round(2))
```

```python
print("===== HANDICRAFT BY STATE =====")

state_counts = handicraft_df["State"].value_counts()

print(state_counts)
```

```python
state_gi = pd.crosstab(
    handicraft_df["State"],
    handicraft_df["GI Status"]
)

state_gi
```

```python
# Normalize GI Status values
handicraft_df["GI Status"] = (
    handicraft_df["GI Status"]
    .str.strip()
    .str.lower()
    .map({
        "yes": "Yes",
        "no": "No"
    })
)

print(handicraft_df["GI Status"].value_counts())
```

```python
print("===== CATEGORIES =====")
print(handicraft_df["Category"].value_counts())
```

```python
print("===== MINISTRY / DEPARTMENT =====")
print(handicraft_df["Ministry/ Department"].value_counts())
```

```python
print("===== SAMPLE CRAFT + REGION DATA =====")

handicraft_df[
    ["State", "District", "Product", "Category", "GI Status"]
].head(30)
```

```python
# Clean GI Status
handicraft_df["GI Status"] = (
    handicraft_df["GI Status"]
    .str.strip()
    .str.lower()
    .map({
        "yes": "Yes",
        "no": "No"
    })
)

# Clean Ministry / Department
handicraft_df["Ministry/ Department"] = (
    handicraft_df["Ministry/ Department"]
    .str.strip()
    .str.lower()
    .replace({
        "dc handicraft": "DC Handicraft",
        "dc handicrafts": "DC Handicraft",
        "dc handloom": "DC Handloom",
        "ministry of msme": "Ministry of MSME",
        "ministry of textiles": "Ministry of Textiles",
        "ministry of commerce and industry.": "Ministry of Commerce and Industry",
    })
)

# Display cleaned distributions
print("===== GI STATUS =====")
print(handicraft_df["GI Status"].value_counts())

print("\n===== MINISTRY / DEPARTMENT =====")
print(handicraft_df["Ministry/ Department"].value_counts())
```

```python
print("Rows:", len(handicraft_df))
print("Columns:", handicraft_df.shape[1])
print("Missing values:", handicraft_df.isnull().sum().sum())
print("Duplicates:", handicraft_df.duplicated().sum())
```

```python
handicraft_df.to_csv(
    "../datasets/processed/odop_handicrafts.csv",
    index=False
)

print("Saved successfully!")
```
